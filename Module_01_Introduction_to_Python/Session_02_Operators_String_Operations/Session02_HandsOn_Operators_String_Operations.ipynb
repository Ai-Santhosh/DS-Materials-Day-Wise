### Step 1 · Arithmetic operators · Retail

```python
# 18% GST on a laptop at Croma
laptop_price = 54990
gst = laptop_price * 18 / 100
total = laptop_price + gst

show(f"Laptop ₹{laptop_price} + GST ₹{gst} = **₹{total}**")
```

```python
# The seven arithmetic operators at a DMart counter
show(f"450 + 120 = {450 + 120}")
show(f"2000 - 1785 = {2000 - 1785}")   # change to return
show(f"3 * 285 = {3 * 285}")           # 3 packs of atta
show(f"1200 / 8 = {1200 / 8}")         # / always gives a float
show(f"50 // 12 = {50 // 12}")         # whole part only
show(f"50 % 12 = {50 % 12}")           # remainder
show(f"2 ** 10 = {2 ** 10}")           # power
```

### Step 2 · Floor division and remainder

```python
# BigBasket packs 50 eggs in trays of 12
eggs, tray_size = 50, 12
show(f"{eggs // tray_size} full trays, {eggs % tray_size} eggs left over")

# An IndiGo flight of 135 minutes in hours and minutes
minutes = 135
show(f"Flight time: {minutes // 60} h {minutes % 60} min")
```

### Step 3 · Assignment operators · Banking

```python
balance = 25000
balance += 42000     # salary credited
balance -= 18500     # home loan EMI
show(f"SBI balance: ₹{balance}")
```

### Step 4 · Comparison operators

```python
cibil_score = 782
show(f"Loan-ready score (>= 750)? {cibil_score >= 750}")
show(f"Exactly 750? {cibil_score == 750}")
show(f"Same bank name? {'SBI' == 'sbi'}")   # case matters
```

### Step 5 · Logical operators · Lending

```python
salary = 48000
score = 782
existing = False
festive_offer = True

eligible = salary >= 25000 and score >= 750
fee_waived = existing or festive_offer

show(f"Loan eligible: {eligible}")
show(f"Processing fee waived: {fee_waived}")
show(f"Send welcome kit: {not existing}")
```

```python
# Brackets change the order
show(f"1000 + 500 * 2 = {1000 + 500 * 2}")
show(f"(1000 + 500) * 2 = {(1000 + 500) * 2}")
```

### Step 9 · Cleaning methods · Healthcare & travel

```python
raw_medicine = "  dolo 650 TABLETS  "
clean = raw_medicine.strip()

show(f"strip: `{clean}`")
show(f"title: `{clean.title()}`")
show(f"upper: `{clean.upper()}`")
show(f"lower: `{clean.lower()}`")
show(f"Original is unchanged: `{raw_medicine!r}`")
```

### Step 10 · Parsing methods · Logistics & beauty
